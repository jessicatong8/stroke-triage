```python
import xgboost as xgb
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV
```

```python
df = pd.read_csv("output/random_10000.csv")
df.head()
```

```python
feature_list = ['sex','age','RACE','time_since_symptoms','time_to_primary','time_to_comprehensive','transfer_time']
y = df['Percent Comprehensive']
X = df[feature_list]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=47)
```

```python
model = xgb.XGBRegressor(
    objective="reg:squarederror",  # Default for regression
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=47
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("R^2:", r2)
```
